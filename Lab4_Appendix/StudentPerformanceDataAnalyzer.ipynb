```python

from collections import Counter, defaultdict
from statistics import mean


# Student-Specific Inputs and Data Generation
LAST_NAME = input("Enter your last name: ").strip().upper()
STUDENT_ID = input("Enter your student ID: ")
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip().upper()

SEED_NUM = int(STUDENT_ID[-1])

if not LAST_NAME or not FAVORITE_ARTIST:
    raise ValueError("Last name and favorite artist cannot be empty.")

if SEED_NUM < 0:
    raise ValueError("Seed number must be non-negative.")

def generate_score(student_index, subject_index):
    # Use the character positions of the favorite artist.
    artist_index = (student_index + subject_index) % len(FAVORITE_ARTIST)
    character_value = ord(FAVORITE_ARTIST[artist_index])

    score = (
        SEED_NUM
        + character_value * (artist_index + 1)
        + (student_index + 1) * 7
        + (subject_index + 1) * 11
    ) % 101

    return score


def generate_records():
    subjects = [
        "PROGRAMMING",
        "MATHEMATICS",
        "ELECTRONICS",
        "PHYSICS",
        "ENGLISH"
    ]

    records = []

    for student_index in range(5):
        student_name = f"{LAST_NAME}_S{student_index + 1}"

        for subject_index, subject in enumerate(subjects):
            score = generate_score(student_index, subject_index)

            records.append({
                "student": student_name,
                "subject": subject,
                "score": score
            })

    records.append(records[0].copy())

    records.append({
        "student": f"{LAST_NAME}_S5",
        "subject": "ENGLISH",
        "score": None
    })

    return records
```

```python
# Organize Records
def organize_records(records):
    student_data = defaultdict(list)
    subject_data = defaultdict(list)

    for record in records:
        student_data[record["student"]].append(record)
        subject_data[record["subject"]].append(record)

    return student_data, subject_data
```

```python
# Determine Performance Category
def get_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 75:
        return "PASSED"
    else:
        return "FAILED"


def categorize_records(records):
    for record in records:
        record["category"] = get_category(record["score"])

    return records
```

```python
# Identify Repeated and Incomplete Records
def analyze_records(records):
    record_counts = Counter(
        (r["student"], r["subject"], r["score"])
        for r in records
    )

    repeated = {
        record: count
        for record, count in record_counts.items()
        if count > 1
    }

    incomplete = [
        record for record in records
        if record["score"] is None
    ]

    return repeated, incomplete
```

```python
# Determine Relevant Performance Summaries
def calculate_summaries(student_data, subject_data):
    student_summary = {}
    subject_summary = {}

    for student, student_records in student_data.items():
        scores = [
            r["score"] for r in student_records
            if r["score"] is not None
        ]

        student_summary[student] = {
            "count": len(scores),
            "average": mean(scores) if scores else 0,
            "highest": max(scores) if scores else None,
            "lowest": min(scores) if scores else None
        }

    for subject, subject_records in subject_data.items():
        scores = [
            r["score"] for r in subject_records
            if r["score"] is not None
        ]

        subject_summary[subject] = {
            "count": len(scores),
            "average": mean(scores) if scores else 0,
            "highest": max(scores) if scores else None,
            "lowest": min(scores) if scores else None
        }

    return student_summary, subject_summary
```

```python
# Arrange Analyzed Results
def display_report(records, student_summary, subject_summary,
                   repeated, incomplete):
    print("\n" + "=" * 65)
    print("STUDENT PERFORMANCE REPORT")
    print("=" * 65)

    print(f"Last Name: {LAST_NAME}")
    print(f"Seed Number: {SEED_NUM}")
    print(f"Favorite Artist: {FAVORITE_ARTIST}")

    print("\n===GENERATED PERFORMANCE RECORDS===")
    for record in records:
        print(record)


    print("\n===STUDENT SUMMARIES===")
    for student, summary in sorted(student_summary.items()):
        print(
            f"{student}: Average={summary['average']:.2f}, "
            f"Highest={summary['highest']}, "
            f"Lowest={summary['lowest']}, "
            f"Valid Scores={summary['count']}"
        )

    print("\n===SUBJECT SUMMARIES===")
    for subject, summary in sorted(subject_summary.items()):
        print(
            f"{subject}: Average={summary['average']:.2f}, "
            f"Highest={summary['highest']}, "
            f"Lowest={summary['lowest']}, "
            f"Valid Scores={summary['count']}"
        )

    print("\n===REPEATED RECORDS===")
    if repeated:
        for record, count in repeated.items():
            print(
                f"Student={record[0]}, Subject={record[1]}, "
                f"Score={record[2]}, Occurrences={count}"
            )
    else:
        print("No repeated records found.")

    print("\n===INCOMPLETE RECORDS===")
    if incomplete:
        for record in incomplete:
            print(record)
    else:
        print("No incomplete records found.")

    print("\n6. PERFORMANCE CATEGORIES")
    category_counts = Counter(
        record["category"] for record in records
    )

    for category, count in sorted(category_counts.items()):
        print(f"{category}: {count}")

    print("\n===ORDERED RESULTS===")
    ordered_records = sorted(
        records,
        key=lambda r: (
            r["student"],
            r["subject"],
            r["score"] is None,
            r["score"] if r["score"] is not None else 0
        )
    )

    for record in ordered_records:
        print(record)

    valid_scores = [
        r["score"] for r in records
        if r["score"] is not None
    ]

    print("\n===FINAL SUMMARY===")
    print(f"Total records: {len(records)}")
    print(f"Valid scores: {len(valid_scores)}")
    print(f"Incomplete records: {len(incomplete)}")
    print(f"Repeated record types: {len(repeated)}")
    print(f"Students: {len(student_summary)}")
    print(f"Subjects: {len(subject_summary)}")

    if valid_scores:
        print(f"Overall average: {mean(valid_scores):.2f}")
```

```python
# Present Consolidated Report
def main():
    records = generate_records()

    records = categorize_records(records)

    repeated, incomplete = analyze_records(records)

    student_data, subject_data = organize_records(records)

    student_summary, subject_summary = calculate_summaries(
        student_data, subject_data
    )

    display_report(
        records,
        student_summary,
        subject_summary,
        repeated,
        incomplete
    )


if __name__ == "__main__":
    main()
```
