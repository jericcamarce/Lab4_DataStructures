```python
from collections import Counter, defaultdict
from statistics import mean

# Student-Specific Inputs and Data Generation
LAST_NAME = input("Enter your last name: ").strip().upper()
STUDENT_ID = input("Enter your student ID: ")
FAVORITE_ARTIST = input("Enter your favorite artist: ").strip().upper()

SEED_NUM = int(STUDENT_ID[-1])

if not LAST_NAME or not FAVORITE_ARTIST:
    raise ValueError("Last name and favorite artist cannot be empty.")

if SEED_NUM < 0:
    raise ValueError("Seed number must be non-negative.")

def generate_value(category, source_index, measurement_index):
    artist_position = (source_index + measurement_index) % len(FAVORITE_ARTIST)
    character_value = ord(FAVORITE_ARTIST[artist_position])
    position = artist_position + 1

    base_values = {
        "TEMPERATURE": 20,
        "VOLTAGE": 5,
        "CURRENT": 1,
        "PRESSURE": 100,
        "SPEED": 10
    }

    ranges = {
        "TEMPERATURE": 40,
        "VOLTAGE": 15,
        "CURRENT": 10,
        "PRESSURE": 900,
        "SPEED": 120
    }

    value = (
        SEED_NUM
        + character_value * position
        + (source_index + 1) * 7
        + measurement_index * 11
    ) % ranges[category]

    return round(base_values[category] + value, 2)

def generate_test_data():
    categories = [
        "TEMPERATURE",
        "VOLTAGE",
        "CURRENT",
        "PRESSURE",
        "SPEED"
    ]

    records = []

    for source_index in range(5):
        source = f"{LAST_NAME}_T{source_index + 1}"

        for category in categories:
            for measurement_index in range(2):
                records.append({
                    "source": source,
                    "category": category,
                    "value": generate_value(
                        category, source_index, measurement_index
                    )
                })

    records.append(records[0].copy())

    return records
```

```python
# Organize Records
def organize_records(records):
    organized = defaultdict(lambda: defaultdict(list))

    for record in records:
        organized[record["source"]][record["category"]].append(
            record["value"]
        )

    return organized
```

```python
# Identify Repeated abd Distinct Data
def analyze_duplicates(records):
    record_counts = Counter(
        (r["source"], r["category"], r["value"])
        for r in records
    )

    repeated = {
        record: count
        for record, count in record_counts.items()
        if count > 1
    }

    distinct = set(record_counts.keys())

    return repeated, distinct
```

```python
# Determine Relevant Numerical and Categorical Summaries
def summarize_data(records):
    category_values = defaultdict(list)
    source_counts = Counter()
    category_counts = Counter()

    for record in records:
        category_values[record["category"]].append(record["value"])
        source_counts[record["source"]] += 1
        category_counts[record["category"]] += 1

    numerical_summary = {}

    for category, values in category_values.items():
        numerical_summary[category] = {
            "count": len(values),
            "mean": mean(values),
            "minimum": min(values),
            "maximum": max(values)
        }

    return numerical_summary, source_counts, category_counts
```

```python
# Arrange Processed Results 
def process_results(records):
    # Sort by source, category, and measurement value.
    return sorted(
        records,
        key=lambda r: (r["source"], r["category"], r["value"])
    )
```

```python
# Present Original and Processed Information
def display_results(records, organized, repeated, distinct,
                    numerical_summary, source_counts, category_counts,
                    processed):
    print("\n" + "=" * 60)
    print("ENGINEERING TEST DATA ORGANIZER")
    print("=" * 60)

    print(f"Student Last Name: {LAST_NAME}")
    print(f"Seed Number: {SEED_NUM}")
    print(f"Favorite Artist: {FAVORITE_ARTIST}")

    print("\n===GENERATED TEST DATA===")
    for record in records:
        print(record)

    print("\n===ORGANIZED TEST DATA===")
    for source, categories in sorted(organized.items()):
        print(f"\nSource: {source}")

        for category, values in sorted(categories.items()):
            print(f"  {category}: {values}")

    print("\n===REPEATED DATA===")
    if repeated:
        for record, count in repeated.items():
            print(
                f"Source={record[0]}, Category={record[1]}, "
                f"Value={record[2]} | Occurrences={count}"
            )
    else:
        print("No repeated records found.")

    print(f"\nDistinct records: {len(distinct)}")
    print(f"Repeated record types: {len(repeated)}")

    print("\n===NUMERICAL SUMMARY===")
    for category, stats in sorted(numerical_summary.items()):
        print(
            f"{category}: Count={stats['count']}, "
            f"Mean={stats['mean']:.2f}, "
            f"Min={stats['minimum']}, "
            f"Max={stats['maximum']}"
        )

    print("\n===CATEGORICAL SUMMARY===")
    print("\nRecords per source:")
    for source, count in sorted(source_counts.items()):
        print(f"{source}: {count}")

    print("\nRecords per category:")
    for category, count in sorted(category_counts.items()):
        print(f"{category}: {count}")

    print("\n ===PROCESSED AND ORDERED RESULTS===")
    for record in processed:
        print(record)

    print("\n ===FINAL SUMMARY===")
    print(f"Total records: {len(records)}")
    print(f"Distinct record types: {len(distinct)}")
    print(f"Repeated record types: {len(repeated)}")
    print(f"Number of test sources: {len(source_counts)}")
    print(f"Number of measurement categories: {len(category_counts)}")
    print(f"Number of unique categories: {len(numerical_summary)}")
```

```python
# Summary
def main():
    records = generate_test_data()

    organized = organize_records(records)

    repeated, distinct = analyze_duplicates(records)

    numerical_summary, source_counts, category_counts = (
        summarize_data(records)
    )

    processed = process_results(records)

    display_results(
        records,
        organized,
        repeated,
        distinct,
        numerical_summary,
        source_counts,
        category_counts,
        processed
    )


if __name__ == "__main__":
    main()
```
