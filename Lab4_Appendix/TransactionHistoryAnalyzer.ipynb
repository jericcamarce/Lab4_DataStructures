```python
def generate_transactions():
    LAST_NAME = input("Enter your last name: ").strip().upper()
    STUDENT_ID = input("Enter your student ID: ")
    FAVORITE_ARTIST = input("Enter your favorite artist: ").strip().upper()

    SEED_NUM = int(STUDENT_ID[-1])

    if not LAST_NAME or not FAVORITE_ARTIST or SEED_NUM < 0:
        raise ValueError("Invalid input.")

    products = {
        "NOTEBOOK": 35.00,
        "PEN": 10.00,
        "USB": 250.00,
        "HEADPHONES": 450.00,
        "CALCULATOR": 350.00
    }

    transactions = []

    for i in range(5):
        customer = f"{LAST_NAME}_C{i + 1}"

        for j, (product, price) in enumerate(products.items()):
            artist_index = (i + j) % len(FAVORITE_ARTIST)
            character = ord(FAVORITE_ARTIST[artist_index])

            quantity = (
                SEED_NUM
                + character * (artist_index + 1)
                + (i + 1) * 7
                + (j + 1) * 11
            ) % 10 + 1

            transactions.append({
                "customer": customer,
                "product": product,
                "quantity": quantity,
                "price": price,
                "total": quantity * price
            })

    transactions.append(transactions[0].copy())

    return transactions, products, LAST_NAME, SEED_NUM, FAVORITE_ARTIST
```

```python
# Organize records
def organize_transactions(transactions):
    customers = defaultdict(list)
    product_data = defaultdict(list)

    for transaction in transactions:
        customers[transaction["customer"]].append(transaction)
        product_data[transaction["product"]].append(transaction)

    return customers, product_data
```

```python
# Determine Total Transaction Values
def calculate_totals(transactions):
    for transaction in transactions:
        transaction["total"] = (
            transaction["quantity"] * transaction["price"]
        )

    return sum(t["total"] for t in transactions)
```

```python
# Identify Repeated Purchases and Distinct Data
def analyze_purchases(transactions):
    purchase_counts = Counter(
        (t["customer"], t["product"])
        for t in transactions
    )

    repeated = {
        purchase: count
        for purchase, count in purchase_counts.items()
        if count > 1
    }

    distinct_customers = set(t["customer"] for t in transactions)
    distinct_products = set(t["product"] for t in transactions)

    return repeated, distinct_customers, distinct_products
```

```python
# Determine Relevant Summaries

def summarize_transactions(customers, product_data):
    customer_summary = {}
    product_summary = {}

    for customer, records in customers.items():
        customer_summary[customer] = {
            "quantity": sum(t["quantity"] for t in records),
            "spent": sum(t["total"] for t in records)
        }

    for product, records in product_data.items():
        product_summary[product] = {
            "quantity": sum(t["quantity"] for t in records),
            "sales": sum(t["total"] for t in records)
        }

    return customer_summary, product_summary
```

```python
# Arrange Analyzed Results
def sort_transactions(transactions):
    return sorted(
        transactions,
        key=lambda t: (t["customer"], t["product"])
    )
```

```python
# Present Consolidated Report
def display_report(transactions, results, student_info):
    last_name, seed_num, favorite_artist = student_info
    customers = results["customers"]
    product_data = results["product_data"]
    total_sales = results["total_sales"]
    repeated = results["repeated"]
    distinct_customers = results["distinct_customers"]
    distinct_products = results["distinct_products"]
    customer_summary = results["customer_summary"]
    product_summary = results["product_summary"]
    ordered = results["ordered"]

    print("\n" + "=" * 65)
    print("TRANSACTION HISTORY ANALYZER")
    print("=" * 65)
    print(f"Last Name: {last_name}")
    print(f"Seed Number: {seed_num}")
    print(f"Favorite Artist: {favorite_artist}")

    print("\n===GENERATED TRANSACTION RECORDS===")
    for transaction in transactions:
        print(transaction)

    print("\n===ORGANIZED TRANSACTIONS BY CUSTOMER===")
    for customer, records in sorted(customers.items()):
        print(f"{customer}: {len(records)} transactions")

    print("\n===TRANSACTION VALUES===")
    for transaction in transactions:
        print(
            f"{transaction['customer']} | {transaction['product']} | "
            f"Quantity: {transaction['quantity']} | "
            f"Total: PHP {transaction['total']:.2f}"
        )
    print(f"Overall transaction value: PHP {total_sales:.2f}")

    print("\n===REPEATED PURCHASES AND DISTINCT DATA===")
    if repeated:
        for (customer, product), count in sorted(repeated.items()):
            print(f"{customer} purchased {product} {count} times.")
    else:
        print("No repeated purchases found.")

    print(f"Distinct customers: {len(distinct_customers)}")
    print(f"Distinct products: {len(distinct_products)}")

    print("\n===CUSTOMER SUMMARIES===")
    for customer, summary in sorted(customer_summary.items()):
        print(
            f"{customer}: Quantity={summary['quantity']}, "
            f"Total Spent=PHP {summary['spent']:.2f}"
        )

    print("\n===PRODUCT SUMMARIES===")
    for product, summary in sorted(product_summary.items()):
        print(
            f"{product}: Quantity Sold={summary['quantity']}, "
            f"Total Sales=PHP {summary['sales']:.2f}"
        )

    print("\n===ORDERED TRANSACTIONS===")
    for transaction in ordered:
        print(transaction)

    print("\n--- OVERALL SUMMARY ---")
    print(f"Total transaction records: {len(transactions)}")
    print(f"Total quantity sold: {sum(t['quantity'] for t in transactions)}")
    print(f"Number of customers: {len(distinct_customers)}")
    print(f"Number of products: {len(distinct_products)}")
    print(f"Overall sales: PHP {total_sales:.2f}")
    print("Transaction analysis completed.")


# MAIN PROGRAM
def main():
    transactions, products, last_name, seed_num, favorite_artist = (
        generate_transactions()
    )

    customers, product_data = organize_transactions(transactions)

    total_sales = calculate_totals(transactions)

    repeated, distinct_customers, distinct_products = (
        analyze_purchases(transactions)
    )

    customer_summary, product_summary = summarize_transactions(
        customers, product_data
    )

    ordered = sort_transactions(transactions)

    results = {
        "customers": customers,
        "product_data": product_data,
        "total_sales": total_sales,
        "repeated": repeated,
        "distinct_customers": distinct_customers,
        "distinct_products": distinct_products,
        "customer_summary": customer_summary,
        "product_summary": product_summary,
        "ordered": ordered
    }

    student_info = (last_name, seed_num, favorite_artist)

    display_report(transactions, results, student_info)


if __name__ == "__main__":
    main()
```
